## Analyze Yearly Delay Trends

### Subtask:
Identify yearly trends for different delay categories.


**Reasoning**:
To identify yearly trends for each delay category, I will iterate through the specified percentage columns, find the year(s) with the highest and lowest percentages for each, and then print these results, as specified in the instructions.



````python
print("\n--- Yearly Delay Trends Analysis ---")

delay_columns = [
    'Air_Carrier_Delay_Pct',
    'Aircraft_Arriving_Late_Pct',
    'National_Aviation_System_Pct',
    'Security_Delay_Pct',
    'Extreme_Weather_Pct',
    'Weather_Total_Share_Pct'
]

for col in delay_columns:
    # Find year(s) with the highest percentage
    max_value = df_cleaned[col].max()
    max_years = df_cleaned[df_cleaned[col] == max_value]['Year'].tolist()
    print(f"\n{col}:")
    print(f"  Highest: {max_value:.2f}% in Year(s): {max_years}")

    # Find year(s) with the lowest percentage
    min_value = df_cleaned[col].min()
    min_years = df_cleaned[df_cleaned[col] == min_value]['Year'].tolist()
    print(f"  Lowest: {min_value:.2f}% in Year(s): {min_years}")
````
